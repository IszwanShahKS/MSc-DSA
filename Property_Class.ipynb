# **Create Class & Inheritance For "Property"**

```python
#########################
##### PARENT CLASSS #####
#########################

class Property:
    
    """Initialize attributes of the Parent Class to describe a property"""
    
    def __init__ (self, type, size, room, bathroom, location, price):
        
        self.type = type
        self.size = size
        self.room = room
        self.bathroom = bathroom
        self.location = location
        self.price = price
    
    def describe_property(self):
        
        """Display a summary of the property"""
        
        property_description = (
            f"Each unit of {self.type} with a built-up area of {self.size:,} sq ft has {self.room} rooms and {self.bathroom} bathrooms."
            f"\nIt is located in {self.location} with a selling price starting from RM {self.price:,}."
        )
        
        print(property_description)
        
    def update_size_price(self, new_size, new_price):
        
        """Update the price of the property due to change in property size"""
        
        self.size = new_size
        self.price = new_price
        
    def update_room_price(self, new_room, new_price):
        
        """Update the price of the property due to change in number of rooms"""
        
        self.room = new_room
        self.price = new_price
        
    def revise_price(self, revise_price):
        
        """Revise the price of the property"""
        
        self.price = revise_price
        
        print(f"The price of this property has been revised and starts from RM {self.price:,} due to the increase in the price of building materials.")
        
    def price_discount(self, price_discount):
        
        """Price discount for the property"""
        
        self.price_discount = price_discount
        self.price -= price_discount
        
        print(f"Year-End-Sales(YES) : Discount worth of RM {self.price_discount:,}. The new price starts from RM {self.price:,}.")


#####################################
### CHILD CLASS 1 (INHERITANCE 1) ###
#####################################

class Landed (Property):
    
    """Initialize attributes of the Parent Class to describe a property"""
    
    def __init__ (self, type, size, room, bathroom, location, price):
        
        """Initialize attributes of the child class to represent aspects of property specific to landed property"""
        
        super().__init__(type, size, room, bathroom, location, price)
        
        self.land = 0

    def extra_land (self,land):

        """State size of extra land for each unit"""
        
        self.land += land

        print(f"Each unit of {self.type} has extra land with a size of {self.land} sq ft.")

#####################################
### CHILD CLASS 2 (INHERITANCE 2) ###
#####################################

class Highrise (Property):
    
    """Initialize attributes of the Parent Class to describe a property"""
    
    def __init__ (self, type, size, room, bathroom, location, price):
        
        """Initialize attributes of the child class to represent aspects of property specific to highrise property"""
        
        super().__init__(type, size, room, bathroom, location, price)
        
        self.floor = 0 # Default level is set to 0
        self.facility = {}
        self.parking = 0 # Default level is set to 0
        
    def total_floor (self,floor):

        """State total floor of the property"""
        
        self.floor += floor
        
        print(f"This {self.type} has {self.floor} floors in total.")

    def add_facilities (self,facility):

        """State facilities at the the property"""
        
        self.facility = facility
        
        print(f"This {self.type} has many facilities such as {', '.join(self.facility)}.")

    def total_parking (self,parking):

        """State total parking for each unit of the property"""
        
        self.parking += parking
        
        print(f"Each unit of {self.type} has {self.parking} parking.")
      
```

# **Instances (Parent Class)**

```python
# Instance 1
# Describing the property

property_1 = Property("'double storey house'", 2480, 6, 4, "Setia Alamsari, Kajang", 650000)
property_1.describe_property()
```

```python
# Instance 2
# Update the price of the property due to change in property size

print(f"[bold]Original Size & Price of The Property:[/bold]\n")
property_2 = Property("'bungalow'", 5600, 7, 5, "Presint 12, Putrajaya", 1250000)
property_2.describe_property()

print(f"\n\n[bold]The Updated Size & Price of The Property:[/bold]\n")
property_2.update_size_price(5700, 1300000)
property_2.describe_property()
```

```python
# Instance 3
# Update the price of the property due to change in number of rooms

print(f"[bold]Original Number of Rooms & Price of The Property:[/bold]\n")
property_3 = Property("'condominium'", 1150, 3, 2, "Gamuda Cove, Sepang", 375000)
property_3.describe_property()

print(f"\n\n[bold]The Updated Number of Rooms  & Price of The Property:[/bold]\n")
property_3.update_room_price(4, 385000)
property_3.describe_property()
```

```python
# Instance 4
# The revise price of the property

print(f"[bold]Property information: [/bold]")
property_1.describe_property()

property_1.revise_price(700000)
```

```python
# Instance 5
# The price discount of the property

print(f"[bold]Property information: [/bold]")
property_2.describe_property()

property_2.price_discount(50000)
```
