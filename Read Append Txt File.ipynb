# **Read Original Files**

```python
path_1 = "Movies.txt"

path_2 = "Sports.txt"

with open (path_1,'r') as file_1, open (path_2,'r') as file_2:
    content1 = file_1.read()
    content2 = file_2.read()
    print("[bold]Movies:[/bold]\n" + content1.strip() + "\n\n" + "[bold]Sports:[/bold]\n" + content2.strip())
```

# **Add New Lines Into Original Files & Read Those Files**

```python
# add new lines into original files

with open (path_1,'a') as file_1, open (path_2,'a') as file_2:
    
    newlines_movies = ["\nThese movies have been nominated in The 96th Academy Awards 2024.\n",
                     "They have competed for the 'Best Picture' Category.\n",
                     "During the award, 'Oppenheimer' has been announced as a winner.\n",
                     "The 96th Academy Awards 2024 has been held in Dolby Theatre at Ovation Hollywood.\n",
                     "It was held to honor movies released in 2023.\n\n"]
    
    # reference: https://www.oscars.org/oscars/ceremonies/2024
    
    newlines_sports = ["\n\nThese sports are The World’s Highest-Paying Sports in 2024.\n",
                     "This list has been released by CEO Today Magazine for its December 2024 issue.\n",
                     "In overall, American sports like basketball and American football dominate this list.\n",
                     "The average annual salary for basketball athlete is $ 10.5 million.\n",
                     "The average annual salary for American football athlete is $ 4.5 million.\n"]

    # reference: https://www.ceotodaymagazine.com/2024/12/the-worlds-highest-paying-sports-leagues-teams-and-salaries/

    file_1.writelines(newlines_movies)
    file_2.writelines(newlines_sports)

# read those files

with open (path_1,'r') as file_1, open (path_2,'r') as file_2:
    content1 = file_1.read()
    content2 = file_2.read()
    print("[bold]Movies:[/bold]\n" + content1.strip() + "\n\n" + "[bold]Sports:[/bold]\n" + content2.strip())
```

# **Read New File With Try-Except Block**

```python
# path_1 and path 2 are as per variables in the "Read Original Files" syntax

try:
    
    with open (path_1,'r') as file_1, open (path_2,'r') as file_2:
        
        content1 = file_1.read()
        content2 = file_2.read()
        print("[bold]Movies:[/bold]\n" + content1.strip() + "\n\n" + "[bold]Sports:[/bold]\n" + content2.strip())

except FileNotFoundError as error:
    
    missing_file = str(error).split("'")[1]
    
    print(f"Sorry. This action can not be perform this time.\n" +
        f"One of your files {missing_file} is not detected.")
```
