# **Price Management System (PMS)**

```python
# Import relevant packages

from rich import print # To formatting text
```

```python
# Create Initial Product list

Product_List = {
    'Bottle Water (500 ml)': [1.30,48],
    'Bottle Water (1500 ml)': [2.50,24],
    'Rice (5 kg)': [15.00,20],
    'Rice (10 kg)': [30.00,10]
    }

print(Product_List)
```

```python
# Define available actions in the PMS

New_Product = "NEW PRODUCT"
Update_Product = "UPDATE PRODUCT"
View_All_Products = "VIEW ALL PRODUCTS"
Stop = "STOP"
Update_Price = "UPDATE PRICE"
Update_Quantity = "UPDATE QUANTITY"
```

```python
# Prompt messages available in the PMS

Prompt_1 = "Hello User!!"
Prompt_2 = "Please choose your action from the list of actions available below:"
Prompt_3 = "1. 'NEW PRODUCT' : User can add a new product together with its price and quantity."
Prompt_4 = "2. 'UPDATE PRODUCT' : User can update existing product's price or quantity."
Prompt_5 = "3. 'VIEW ALL PRODUCTS' : User can view all products that has been store in the PMS."
Prompt_6 = "4. 'STOP' : User intends to stop and exit from the PMS."
Prompt_7 = "1. 'UPDATE PRICE' : User can update price of existing product in the PMS."
Prompt_8 = "2. 'UPDATE QUANTITY' : User can update quantity of existing product in the PMS."
```

```python
### ### PMS PROGRAM ###
###################

#-----------------------------------------------------------------------------------------------#
### Show prompt message to let user choose available actions ###


print(Prompt_1 + ("\n") + Prompt_2 + ("\n\n") + Prompt_3 + ("\n") +
      Prompt_4 + ("\n") + Prompt_5 + ("\n") + Prompt_6 + ("\n"))


#-----------------------------------------------------------------------------------------------#
### Using 'while loop' to let user choose his/her required action & do neccessary actions ##


while True:
    action = input("Please Enter Your Action.\n(This space is case-sensitive.)\n")


######-------------------------------------------------------------######
### Action 1 : Add a new product together with its price and quantity ###
######-------------------------------------------------------------######


    if action == New_Product:

        print("[bold]\n\nPlease add new product in the PMS:[/bold]")

        # Function to add new product into the PMS

        def add_new_product(Product_List):

            # Enter Product Name

            Product_Name = input("\nEnter Product Name:\nPlease follow this format:(Ex: 'Bottle Water (500 ml)')\n").strip()

            # check whether product already exist in the Product List

            if Product_Name in Product_List:
                print(f"\nProduct '{Product_Name}' existed in the PMS.")

            else:

            # Add price & quantity if it is new product

                try:
                    Product_Price = float(input("\nEnter Product Price (RM):\n(value up to 2 decimal points)\n"))
                    Product_Quantity = int(input("\nEnter Quantity (in unit):\n(value without decimal points)\n"))

                # show remark to notify user about error in input

                except ValueError:
                    print("Error: Ensure 'Product Price' & 'Product Quantity' are following requirements stated.")
                    return # exit function since input for 'Product Price' & 'Product Quantity' are invalid

                # add new product in the Product List

                Product_List[Product_Name] = [round(Product_Price, 2), Product_Quantity]

                # show the success of adding new products with an updated Product List

                print(f"\n\nProduct '{Product_Name}' has been added successfully!")

                print("[bold]\n\nUPDATED PRODUCTS LIST:\n[/bold]")

                for product, details in Product_List.items():
                    price, quantity = details
                    print(f"\nProduct Name: {product}")
                    print(f"Price: RM {price:.2f}") # price with 2 decimal point
                    print(f"Quantity (Unit): {quantity} \n")

        # run function
        add_new_product(Product_List)

        break

    ######-------------------------------------------------------------######
    ######## Action 2 : Update existing product's price or quantity #########
    ######-------------------------------------------------------------######

    elif action == Update_Product:

        print("[bold]\n\nPlease enter 'Product Name' to update its price or quantity:[/bold]")

        # Enter Product Name

        Product_Name = input("\nEnter Product Name:\nPlease follow this format:(Ex: Bottle Water (500 ml))\n(Caution: This space is case-sensitive.)\n").strip()

        # check whether product already exist in the Product List

        if Product_Name in Product_List:

            print(f"\n\nProduct '{Product_Name}' existed in the PMS.")
            print(("\n\n") + Prompt_2 + ("\n") + Prompt_7 + ("\n") + Prompt_8)

            while True:
            
                action = input("\n\nPlease Enter Your Action.\n(This space is case-sensitive.)\n")

                ######-------------------------------------------------------------######
                ############# Action 2(i): Update existing product's price ##############
                ######-------------------------------------------------------------######

                if action == Update_Price:

                    # Function to update product's price

                    def update_price(Product_List):

                        while True: # the process will be recurring until right requirement is fulfilled
                            try:
                                Product_Price = float(input("\nEnter Product Price (RM):\n(value up to 2 decimal points)\n"))

                                # update Product Price in the Product List

                                Product_List[Product_Name][0] = round(Product_Price, 2)

                                break

                            # show remark to notify user about error in input

                            except ValueError:
                                print("Error: Ensure 'Product Price' are following requirements stated.")

                        # show the success of updating product price with an updated Product List

                        print(f"\n\nPrice of product '{Product_Name}' has been updated successfully to RM {Product_Price:.2f}!")

                        print("[bold]\n\nUPDATED PRODUCTS LIST:\n[/bold]")

                        for product, details in Product_List.items():
                            price, quantity = details
                            print(f"\nProduct Name: {product}")
                            print(f"Price: RM {price:.2f}") # price with 2 decimal point
                            print(f"Quantity (Unit): {quantity} \n")

                    # run function

                    update_price(Product_List)

                    break

                ######-------------------------------------------------------------######
                ############# Action 2(ii): Update existing product's price #############
                ######-------------------------------------------------------------######

                elif action == Update_Quantity:

                    # Function to update product's quantity

                    def update_quantity(Product_List):

                        while True: # the process will be recurring until right requirement is fulfilled
                            try:
                                Product_Quantity = int(input("\nEnter Quantity (in unit):\n(value without decimal points)\n"))

                                # update Product Quantity in the Product List

                                Product_List[Product_Name][1] = Product_Quantity

                                break

                            # show remark to notify user about error in input

                            except ValueError:
                                print("Error: Ensure 'Product Quantity' are following requirements stated.")

                        # show the success of updating product quantity with an updated Product List

                        print(f"\n\nQuantity of product '{Product_Name}' has been updated successfully to {Product_Quantity}!")

                        print("[bold]\n\nUPDATED PRODUCTS LIST:\n[/bold]")

                        for product, details in Product_List.items():
                            price, quantity = details
                            print(f"\nProduct Name: {product}")
                            print(f"Price: RM {price:.2f}") # price with 2 decimal point
                            print(f"Quantity (Unit): {quantity} \n")

                    # run function

                    update_quantity(Product_List)

                    break

                ######-------------------------------------------------------------#####
                ################### Ask user to enter correct action ###################
                ######-------------------------------------------------------------#####

                else:

                    print("[bold]\n\nPlease choose again the action you require from the list of available actions as stated above.")

        else:

            print(f"\nProduct '{Product_Name}' does not exist in the PMS.")

            break

        break


    ######-------------------------------------------------------------######
    ###### Action 3 : View all products that has been store in the PMS ######
    ######-------------------------------------------------------------######


    elif action == View_All_Products:

        # print all available products together with their prices & quantities that has been stored in the PMS
        # using 'for loop'

        print("[bold]\n\nPRODUCTS LIST:\n[/bold]")

        for product, details in Product_List.items():
            price, quantity = details
            print(f"\nProduct Name: {product}")
            print(f"Price: RM {price:.2f}") # price with 2 decimal point
            print(f"Quantity (Unit): {quantity} \n")

        break


######-------------------------------------------------------------######
# Action 4 : Stop program if user intends to stop and exit from the PMS #
######-------------------------------------------------------------######


    elif action == Stop:

        print("[bold]\n\nYou intend to stop and exit from using the Price Management System (PMS).[/bold]" +
              "[bold]\nThis program stop here.[/bold]" +
              "[bold]\nThank you for your time.[/bold]")

        break


######-------------------------------------------------------------######
############# Let user choose again his/her requires action #############
######### since early action does not meet any available actions ########
######-------------------------------------------------------------######


    else:

        print("[bold]\n\nPlease choose again the action you require from the list of available actions as stated above.")
```
